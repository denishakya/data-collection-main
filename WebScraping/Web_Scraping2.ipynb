```python
import numpy as np
import pandas as pd
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
```

### Open the Website

```python
driver = webdriver.Chrome()
driver.get("https://www.imdb.com/chart/top/")
```

```python
# Title
# /html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/div[2]/a/h3
# /html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[250]/div/div/div/div/div[2]/div[2]/a/h3

# Year
# /html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/div[3]/span[1]

# Runtime
# /html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/div[3]/span[2]

# Rating
# /html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/span/div/span/span[1]

# Votes
# /html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/span/div/span/span[2]

# Certification
# /html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/div[3]/span[3]

# Image
# /html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[1]/div/div[1]/img
```

### Extract the data

```python
title = []
year = []
runtime = []
imdb_rate = []
votes = []
certification = []
img_url = []

for i in range(1,251):
    
    # Title 
    try:
        movie_name = driver.find_element(By.XPATH,f'/html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/div[2]/a/h3').text
        title.append(movie_name)
    except:
        title.append(np.nan)

    # Year
    try:
        release_year = driver.find_element(By.XPATH,f'/html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/div[3]/span[1]').text
        year.append(release_year)
    except:
        year.append(np.nan)

    # Runtime 
    try:
        duration = driver.find_element(By.XPATH,f'/html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/div[3]/span[2]').text
        runtime.append(duration)
    except:
        runtime.append(np.nan)

    # Rating 
    try:
        rating = driver.find_element(By.XPATH,f'/html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/span/div/span/span[1]').text
        imdb_rate.append(rating)
    except:
        imdb_rate.append(np.nan)

    # Votes
    try:
        vote = driver.find_element(By.XPATH,f'/html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/span/div/span/span[2]').text
        votes.append(vote)
    except:
        votes.append(np.nan)

    # Certification
    try:
        certify = driver.find_element(By.XPATH,f'/html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[2]/div[3]/span[3]').text
        certification.append(certify)
    except:
        certification.append(np.nan)
    
    # Image URL
    try:
        img = driver.find_element(By.XPATH,f'/html/body/div[2]/main/div/div[2]/section/div/div[2]/div/ul/li[{i}]/div/div/div/div/div[1]/div/div[1]/img').get_attribute('src')
        img_url.append(img)
    except:
        img_url.append(np.nan)

print(" Extracted data Sucessfully ")
```

```python
driver.quit()
```

```python
movie_dict = {
    "title":title,
    "year":year,
    "runtime":runtime,
    "imdb_rate":imdb_rate,
    "votes":votes,
    "certification":certification,
    "img_url":img_url
}
```
