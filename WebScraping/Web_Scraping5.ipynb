### Creating Custom Function

```python
def create_df(data):
    columns = data.split('\n')[0].split(" ")
    cols = columns[:-2] + [' '.join(columns[-1:-3:-1][::-1])]
    data_list = data.split("\n")[1:]
    arr = np.array(data_list)
    df = pd.DataFrame(arr.reshape((int(arr.shape[0] / 8)), 8))
    
    df[cols[2:10]] = df[2].str.split(expand = True)
    df.drop(columns = 2,inplace=True)
    
    recent_form = ['-'.join(item) for item in np.array(df.iloc[:, 2:7])]
    df['recent_form'] = recent_form
    df.drop(columns = [i for i in range(3,8)],inplace = True)
    
    df.columns = cols
    
    for i in cols[2:6]:
        df[i] = df[i].astype(int)
    df['NRR'] = df['NRR'].apply(lambda x : float(x))
    df['PTS'] = df['PTS'].apply(lambda x : int(x))
    
    return df
```

```python
def ipl_table(year):
    
    url = f"https://www.iplt20.com/points-table/men/{year}"
    
    driver = webdriver.Chrome()
    driver.get(url)

    time.sleep(10)

    table_xpath = '/html/body/div[1]/section/div/div/div[1]/div/div/div[1]/div/div/div[2]/div/div/div/table'
    data = driver.find_element(By.XPATH, table_xpath).text

    driver.quit()

    return create_df(data)
```
