```python
import numpy as np
import pandas as pd
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
```

### Open the Website

```python
driver = webdriver.Chrome()
driver.get("https://www.iplt20.com/stats/2025")
```

```python
import numpy as np
import pandas as pd

POS = []
Player = []
GT = []
Runs = []
Mat = []
Inns = []
NO = []
HS = []
Avg = []
BF = []
SR = []
hundreds = []
fifties = []
fours = []
sixes = []

for i in range(1,158): 
    
    # -------- POSITION --------
    try:
        pos_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[1]").text
        POS.append(pos_val)
    except:
        POS.append(np.nan)
    
    # -------- PLAYER NAME --------
    try:
        player_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[2]/div/a/div[1]").text
        Player.append(player_val)
    except:
        Player.append(np.nan)
    
    # -------- TEAM --------
    try:
        team_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[2]/div/a/div[2]").text
        GT.append(team_val)
    except:
        GT.append(np.nan)
    
    # -------- RUNS SCORED --------
    try:
        runs_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[3]").text
        Runs.append(runs_val)
    except:
        Runs.append(np.nan)
    
    # -------- MATCHES --------
    try:
        mat_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[4]").text
        Mat.append(mat_val)
    except:
        Mat.append(np.nan)
    
    # -------- INNINGS --------
    try:
        inns_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[5]").text
        Inns.append(inns_val)
    except:
        Inns.append(np.nan)
    
    # -------- NOT OUTS --------
    try:
        no_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[6]").text
        NO.append(no_val)
    except:
        NO.append(np.nan)
    
    # -------- HIGHEST SCORE --------
    try:
        hs_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[7]").text
        HS.append(hs_val)
    except:
        HS.append(np.nan)
    
    # -------- BATTING AVERAGE --------
    try:
        avg_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[8]").text
        Avg.append(avg_val)
    except:
        Avg.append(np.nan)
    
    # -------- BALLS FACED --------
    try:
        bf_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[9]").text
        BF.append(bf_val)
    except:
        BF.append(np.nan)
    
    # -------- STRIKE RATE --------
    try:
        sr_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[10]").text
        SR.append(sr_val)
    except:
        SR.append(np.nan)
    
    # -------- CENTURIES --------
    try:
        hundreds_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[11]").text
        hundreds.append(hundreds_val)
    except:
        hundreds.append(np.nan)
    
    # -------- FIFTIES --------
    try:
        fifties_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[12]").text
        fifties.append(fifties_val)
    except:
        fifties.append(np.nan)
    
    # -------- NUMBER OF FOURS --------
    try:
        fours_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[13]").text
        fours.append(fours_val)
    except:
        fours.append(np.nan)
    
    # -------- NUMBER OF SIXES --------
    try:
        sixes_val = driver.find_element(By.XPATH, f"/html/body/div[1]/div[2]/div[4]/div/section/div/div[3]/div[2]/div[2]/div/div[3]/div/div[1]/table/tbody/tr[{i}]/td[14]").text
        sixes.append(sixes_val)
    except:
        sixes.append(np.nan)

print("!! Data successfully extracted !!")

ipl_dict={'POS':POS,'Player':Player,'GT':GT,'Runs':Runs,
          'Mat':Mat,'Inns':Inns,'NO':NO,'HS':HS,
          'Avg':Avg,'BF':BF,'SR':SR,'hundreds':hundreds,
          'fifties':fifties,'fours':fours,'sixes':sixes}

df_ipl = pd.DataFrame(ipl_dict)
df_ipl.to_csv(f'ipl_data.csv',index=False)
```

### Converting the DataType into integer and float

```python
df_ipl["HS"] = df_ipl["HS"].str.replace('*','')
```

```python
df_ipl["Avg"]=df_ipl["Avg"].str.replace('-','0')
```

```python
list1 = ["Runs", "Mat", "Inns", "NO", "HS", "BF", "hundreds", "fifties", "fours", "sixes"]
for item in list1:
    df_ipl[item] = df_ipl[item].astype(int)

list2 = ["Avg", "SR"]
for item in list2:
    df_ipl[item] = df_ipl[item].astype(float)
```
