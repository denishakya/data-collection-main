```python
import numpy as np
import pandas as pd
```

```python
from selenium import webdriver   # Website : open
from selenium.webdriver.common.by import By     # Locate the element
from selenium.webdriver.common.keys import Keys
```

### Open the Website

```python
driver = webdriver.Chrome()
driver.get("https://sipalaya.com/all-courses-2/")
```

### Extract the data

```python
title = []
course_type = []
duration = []
old_price = []
new_price = []
img_url = []

for i in range(1,14):
    
    # Title 
    try:
        course_name = driver.find_element(By.XPATH,f'/html/body/div[1]/div/div[1]/div/div/div[2]/div/div[{i}]/div[3]/h3/a').text    
        title.append(course_name)
    except:
        title.append(np.nan)

    # Course Type
    try:
        ctype = driver.find_element(By.XPATH,f'/html/body/div[1]/div/div[1]/div/div/div[2]/div/div[{i}]/div[3]/div[3]/div[2]/a[2]').text
        course_type.append(ctype)
    except:
        course_type.append(np.nan)

    # Duration 
    try:
        dur = driver.find_element(By.XPATH,f'/html/body/div[1]/div/div[1]/div/div/div[2]/div/div[{i}]/div[3]/div[2]/div/span[2]').text
        duration.append(dur)
    except:
        duration.append(np.nan)

    # Old Price 
    try:
        op = driver.find_element(By.XPATH,f'/html/body/div[1]/div/div[1]/div/div/div[2]/div/div[{i}]/div[4]/div/div[1]/span/del/span/bdi').text
        old_price.append(op)
    except:
        old_price.append(np.nan)

    # New Price 
    try:
        np = driver.find_element(By.XPATH,f'/html/body/div[1]/div/div[1]/div/div/div[2]/div/div[{i}]/div[4]/div/div[1]/span/ins/span/bdi').text
        new_price.append(np)
    except:
        new_price.append(np.nan)

    # Image URL
    try:
        img = driver.find_element(By.XPATH,f'/html/body/div[1]/div/div[1]/div/div/div[2]/div/div[{i}]/div[1]/a/div/img').get_attribute('src')
        img_url.append(img)
    except:
        img_url.append(np.nan)

print(" Extracted data Sucessfully ")
```

```python
driver.quit()
```

```python
mydict = {"title":title, "course_type":course_type, "duration":duration, "old_price":old_price, "new_price":new_price, "img_url":img_url}
```

```python
dfs = pd.DataFrame(mydict)
dfs
```
