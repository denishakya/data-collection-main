```python
links = []
total_jobs = []

driver = webdriver.Chrome()
driver.get("https://merojob.com/category")

time.sleep(4)
driver.execute_script("window.scrollBy(0, document.body.scrollHeight);")
time.sleep(3)
driver.execute_script("window.scrollTo(0, 0)")
time.sleep(3)

for i in range(1,21):
    try:
        link = driver.find_element(By.XPATH, f'/html/body/div/div/main/div/div[2]/div/div[2]/div[2]/div[2]/a[{i}]').get_attribute('href')
        links.append(link)
    except Exception as e:
        print(e)

    try:
        total = driver.find_element(By.XPATH, f'/html/body/div/div/main/div/div[2]/div/div[2]/div[2]/div[2]/a[{i}]/div/span[2]').text
        total_jobs.append(int(total[1:-1]))
    except Exception as e:
        print(e)

driver.quit()

print("!! Link Extracted Successfully !!")
```

```python
import time
import numpy as np
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
import math

def extract_data(url, total):

    driver = webdriver.Chrome()

    job_title = []
    company = []
    level = []
    employment_type = []
    location = []
    experience_required = []
    salary = []
    skills_required = []
    deadline = []

    category = url.split('/')[-1]
    limit = 20
    total_pages = math.ceil(total / limit)

    heading = category.replace("-", " ").title()
    print(heading)
    
    try:
        for page in range(1, total_pages+1):
            links = url + f'?limit={limit}&offset={page}'
            
            driver.get(links)

            time.sleep(5)
            driver.execute_script("window.scrollBy(0, document.body.scrollHeight);")
            time.sleep(3)
            driver.execute_script("window.scrollTo(0, 0)")
            time.sleep(2)

            for i in range(1, limit+1):

                # Job Title
                try:
                    t = driver.find_element(By.XPATH, f"/html/body/div/div/main/div/div[2]/div/div[2]/div/div[2]/div[3]/div[1]/div[{i}]/div/div[1]/div/div[2]/div/h3/a/div/div/span").text
                    job_title.append(t)
                except:
                    job_title.append(np.nan)

                # Company
                try:
                    c = driver.find_element(By.XPATH, f"/html/body/div/div/main/div/div[2]/div/div[2]/div/div[2]/div[3]/div[1]/div[{i}]/div/div[1]/div/div[2]/div/p/a/span/span").text
                    company.append(c)
                except:
                    company.append(np.nan)

                # Level
                try:
                    l = driver.find_element(By.XPATH, f"/html/body/div/div/main/div/div[2]/div/div[2]/div/div[2]/div[3]/div[1]/div[{i}]/div/div[2]/div[1]/div[1]/p[2]").text
                    level.append(l)
                except:
                    level.append(np.nan)

                # Employment Type
                try:
                    et = driver.find_element(By.XPATH, f"/html/body/div/div/main/div/div[2]/div/div[2]/div/div[2]/div[3]/div[1]/div[{i}]/div/div[2]/div[1]/div[3]/p[2]").text
                    employment_type.append(et)
                except:
                    employment_type.append(np.nan)

                # Location
                try:
                    loc = driver.find_element(By.XPATH, f"/html/body/div/div/main/div/div[2]/div/div[2]/div/div[2]/div[3]/div[1]/div[{i}]/div/div[2]/div[1]/div[5]/span/span").text
                    location.append(loc)
                except:
                    location.append(np.nan)

                # Experience Required
                try:
                    exp = driver.find_element(By.XPATH, f"/html/body/div/div/main/div/div[2]/div/div[2]/div/div[2]/div[3]/div[1]/div[{i}]/div/div[2]/div[2]/div[1]/p[2]").text
                    experience_required.append(exp)
                except:
                    experience_required.append(np.nan)

                # Salary
                try:
                    sal = driver.find_element(By.XPATH, f"/html/body/div/div/main/div/div[2]/div/div[2]/div/div[2]/div[3]/div[1]/div[{i}]/div/div[2]/div[2]/div[3]/p[2]").text
                    salary.append(sal)
                except:
                    salary.append(np.nan)

                # Skills Required
                try:
                    skl = driver.find_element(By.XPATH, f"/html/body/div/div/main/div/div[2]/div/div[2]/div/div[2]/div[3]/div[1]/div[{i}]/div/div[2]/div[3]").text
                    skills_required.append(skl)
                except:
                    skills_required.append(np.nan)

                # Deadline
                try:
                    dl = driver.find_element(By.XPATH, f"/html/body/div/div/main/div/div[2]/div/div[2]/div/div[2]/div[3]/div[1]/div[{i}]/div/div[3]").text
                    deadline.append(dl)
                except:
                    deadline.append(np.nan)

            print(f"--> {page} page extracted successfully.")

        driver.quit()

        # Create dictionary using ALL lists
        data_dict = {
            "job_title": job_title,
            "company": company,
            "level": level,
            "employment_type": employment_type,
            "location": location,
            "experience_required": experience_required,
            "salary": salary,
            "skills_required": skills_required,
            "deadline": deadline
        }

        df_job = pd.DataFrame(data_dict)

        df_job = df_job.dropna(how='all')
        
        df_job.to_csv(f"C:\\Users\\hp\\Desktop\\Jupyter Notebook\\WebScraping\\MeroJob\\{category}.csv", index=False)

        print("Data Extraction Completed.\n")
        return df_job

    except Exception as e:
        print("Error:", e)
```

```python
final = []

for link, total_job in zip(links, total_jobs):
    final.append(extract_data(link, total_job))
```
